```python
import os
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"

import json
import yaml
import joblib
import numpy as np
import pandas as pd

from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
from tensorflow.keras.callbacks import EarlyStopping

# ===============================
# PATHS (DO NOT CHANGE)
# ===============================
BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Tax Revenue Forecasting System"
DATA_PATH = os.path.join(BASE_DIR, "D48-VATandGST_24.csv")

MODEL_PATH = os.path.join(BASE_DIR, "tax_forecast_model.h5")
SCALER_PATH = os.path.join(BASE_DIR, "scaler.pkl")
CONFIG_PATH = os.path.join(BASE_DIR, "config.yaml")
PRED_PATH = os.path.join(BASE_DIR, "predictions.json")

# ===============================
# LOAD DATA
# ===============================
df = pd.read_csv(DATA_PATH)

# Drop junk columns
df = df.loc[:, ~df.columns.str.contains("^Unnamed", na=False)]

# Rename safely (use only first 3 real columns)
df = df.iloc[:, :3]
df.columns = ["city", "year", "revenue"]

# ===============================
# CLEAN YEAR COLUMN (FIX)
# ===============================
# Convert to string → extract 4-digit year → numeric
df["year"] = (
    df["year"]
    .astype(str)
    .str.extract(r"(\d{4})")[0]
)

df["year"] = pd.to_numeric(df["year"], errors="coerce")

# Drop rows with invalid year or revenue
df = df.dropna(subset=["year", "revenue"])

df["year"] = df["year"].astype(int)

# ===============================
# SORT & AGGREGATE
# ===============================
df = df.sort_values(["city", "year"])

# Aggregate city-wise revenue per year
yearly_revenue = (
    df.groupby("year", as_index=False)["revenue"]
    .sum()
)

# ===============================
# SCALING
# ===============================
scaler = MinMaxScaler()
scaled_revenue = scaler.fit_transform(
    yearly_revenue[["revenue"]]
)

joblib.dump(scaler, SCALER_PATH)

# ===============================
# CREATE SEQUENCES
# ===============================
def create_sequences(data, window):
    X, y = [], []
    for i in range(len(data) - window):
        X.append(data[i:i + window])
        y.append(data[i + window])
    return np.array(X), np.array(y)

WINDOW_SIZE = 3
X, y = create_sequences(scaled_revenue, WINDOW_SIZE)

X = X.reshape((X.shape[0], X.shape[1], 1))

# ===============================
# LSTM MODEL
# ===============================
model = Sequential([
    LSTM(64, input_shape=(WINDOW_SIZE, 1)),
    Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mse"
)

early_stop = EarlyStopping(
    monitor="loss",
    patience=10,
    restore_best_weights=True
)

model.fit(
    X, y,
    epochs=100,
    batch_size=8,
    callbacks=[early_stop],
    verbose=1
)

model.save(MODEL_PATH)

# ===============================
# FUTURE FORECAST (NEXT 5 YEARS)
# ===============================
future_steps = 5
last_sequence = scaled_revenue[-WINDOW_SIZE:]
last_sequence = last_sequence.reshape(1, WINDOW_SIZE, 1)

future_scaled = []

for _ in range(future_steps):
    next_val = model.predict(last_sequence, verbose=0)[0][0]
    future_scaled.append(next_val)
    last_sequence = np.append(
        last_sequence[:, 1:, :],
        [[[next_val]]],
        axis=1
    )

future_revenue = scaler.inverse_transform(
    np.array(future_scaled).reshape(-1, 1)
).flatten()

future_years = list(
    range(
        int(yearly_revenue["year"].max()) + 1,
        int(yearly_revenue["year"].max()) + 1 + future_steps
    )
)

prediction_output = {
    "forecast_horizon_years": future_steps,
    "predictions": {
        str(year): float(value)
        for year, value in zip(future_years, future_revenue)
    }
}

with open(PRED_PATH, "w") as f:
    json.dump(prediction_output, f, indent=4)

# ===============================
# SAVE CONFIG
# ===============================
config = {
    "model": "LSTM",
    "window_size": WINDOW_SIZE,
    "optimizer": "adam",
    "loss": "mse",
    "forecast_years": future_steps,
    "scaler": "MinMaxScaler",
    "dataset": "D48-VATandGST_24.csv"
}

with open(CONFIG_PATH, "w") as f:
    yaml.dump(config, f)

print("\n✅ PIPELINE COMPLETED SUCCESSFULLY")
print("📁 Saved in:", BASE_DIR)
```
