```python
import os
import yaml
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error

# ===============================
# PATHS
# ===============================
BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Tax Revenue Forecasting System"
DATA_PATH = os.path.join(BASE_DIR, "D48-VATandGST_24.csv")

RESULT_CSV = os.path.join(BASE_DIR, "hybrid_model_results.csv")
FUTURE_CSV = os.path.join(BASE_DIR, "hybrid_future_predictions.csv")

COMPARE_PLOT = os.path.join(BASE_DIR, "hybrid_actual_vs_predicted.png")
RESULT_PLOT = os.path.join(BASE_DIR, "hybrid_results_trend.png")
FORECAST_PLOT = os.path.join(BASE_DIR, "hybrid_future_forecast.png")
HEATMAP_PLOT = os.path.join(BASE_DIR, "hybrid_revenue_heatmap.png")

CONFIG_PATH = os.path.join(BASE_DIR, "hybrid_config.yaml")

# ===============================
# LOAD & CLEAN DATA
# ===============================
df = pd.read_csv(DATA_PATH)
df = df.loc[:, ~df.columns.str.contains("^Unnamed", na=False)]
df = df.iloc[:, :3]
df.columns = ["city", "year", "revenue"]

df["year"] = (
    df["year"].astype(str)
    .str.extract(r"(\d{4})")[0]
)
df["year"] = pd.to_numeric(df["year"], errors="coerce")
df = df.dropna(subset=["year", "revenue"])
df["year"] = df["year"].astype(int)

df = df.sort_values(["city", "year"])

# ===============================
# HEATMAP (SHOW + SAVE)
# ===============================
pivot = df.pivot_table(values="revenue", index="city", columns="year", aggfunc="sum")
plt.figure(figsize=(14, 8))
sns.heatmap(pivot, cmap="YlGnBu")
plt.title("Hybrid AIS–CSA | City-wise Tax Revenue Heatmap")
plt.tight_layout()
plt.show()
plt.savefig(HEATMAP_PLOT)
plt.close()

# ===============================
# YEARLY AGGREGATION
# ===============================
yearly = df.groupby("year", as_index=False)["revenue"].sum()

X = yearly["year"].values.reshape(-1, 1)
y = yearly["revenue"].values.reshape(-1, 1)

# ===============================
# SCALE TARGET
# ===============================
scaler = MinMaxScaler()
y_scaled = scaler.fit_transform(y)

# ===============================
# FITNESS FUNCTION
# ===============================
def fitness(params):
    w, b = params
    y_pred = w * X + b
    return mean_squared_error(y_scaled, y_pred)

# ===============================
# AIS INITIALIZATION
# ===============================
np.random.seed(42)
POP_SIZE = 20
ais_population = np.random.uniform(-1, 1, size=(POP_SIZE, 2))
ais_fitness = np.array([fitness(ind) for ind in ais_population])

# ===============================
# AIS CLONAL SELECTION
# ===============================
for _ in range(20):
    best_idx = np.argsort(ais_fitness)[:5]
    clones = ais_population[best_idx] + np.random.normal(0, 0.1, (5, 2))
    clone_fitness = np.array([fitness(c) for c in clones])

    combined = np.vstack([ais_population, clones])
    combined_fit = np.hstack([ais_fitness, clone_fitness])

    top = np.argsort(combined_fit)[:POP_SIZE]
    ais_population = combined[top]
    ais_fitness = combined_fit[top]

# ===============================
# CSA REFINEMENT
# ===============================
best_solution = ais_population[np.argmin(ais_fitness)]

for _ in range(30):
    new_solution = best_solution + np.random.uniform(-0.05, 0.05, 2)
    if fitness(new_solution) < fitness(best_solution):
        best_solution = new_solution

w_opt, b_opt = best_solution

# ===============================
# FINAL PREDICTIONS
# ===============================
y_pred_scaled = w_opt * X + b_opt
y_pred = scaler.inverse_transform(y_pred_scaled)

# ===============================
# RESULTS CSV
# ===============================
results_df = pd.DataFrame({
    "year": yearly["year"],
    "actual_revenue": y.flatten(),
    "predicted_revenue": y_pred.flatten()
})
results_df.to_csv(RESULT_CSV, index=False)

# ===============================
# ACTUAL VS PREDICTED GRAPH
# ===============================
plt.figure(figsize=(10, 5))
plt.plot(results_df["year"], results_df["actual_revenue"], label="Actual", marker="o")
plt.plot(results_df["year"], results_df["predicted_revenue"], label="Hybrid Predicted", marker="o")
plt.title("Hybrid AIS–CSA | Actual vs Predicted Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.legend()
plt.show()
plt.savefig(COMPARE_PLOT)
plt.close()

# ===============================
# RESULTS TREND GRAPH
# ===============================
plt.figure(figsize=(10, 5))
plt.plot(results_df["year"], results_df["predicted_revenue"], marker="o")
plt.title("Hybrid AIS–CSA | Revenue Trend")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.show()
plt.savefig(RESULT_PLOT)
plt.close()

# ===============================
# FUTURE FORECAST
# ===============================
future_years = np.arange(
    yearly["year"].max() + 1,
    yearly["year"].max() + 6
).reshape(-1, 1)

future_scaled = w_opt * future_years + b_opt
future_values = scaler.inverse_transform(future_scaled)

future_df = pd.DataFrame({
    "year": future_years.flatten(),
    "forecast_revenue": future_values.flatten()
})
future_df.to_csv(FUTURE_CSV, index=False)

# ===============================
# FUTURE FORECAST GRAPH
# ===============================
plt.figure(figsize=(8, 4))
plt.plot(future_df["year"], future_df["forecast_revenue"], marker="o")
plt.title("Hybrid AIS–CSA | Future Tax Revenue Forecast")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.show()
plt.savefig(FORECAST_PLOT)
plt.close()

# ===============================
# SAVE CONFIG
# ===============================
config = {
    "model": "Hybrid AIS + CSA Optimized Regression",
    "population_size": POP_SIZE,
    "ais_iterations": 20,
    "csa_iterations": 30,
    "optimized_weights": {
        "slope": float(w_opt),
        "intercept": float(b_opt)
    }
}

with open(CONFIG_PATH, "w") as f:
    yaml.dump(config, f)

print("✅ HYBRID AIS + CSA PIPELINE COMPLETED")
print("📁 All outputs saved with 'hybrid_' prefix")
```
