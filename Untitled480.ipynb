```python
import os
import yaml
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error

# ===============================
# PATHS
# ===============================
BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Tax Revenue Forecasting System"
DATA_PATH = os.path.join(BASE_DIR, "D48-VATandGST_24.csv")

RESULT_CSV = os.path.join(BASE_DIR, "his_model_results.csv")
FUTURE_CSV = os.path.join(BASE_DIR, "his_future_predictions.csv")

COMPARE_PLOT = os.path.join(BASE_DIR, "his_actual_vs_predicted.png")
RESULT_PLOT = os.path.join(BASE_DIR, "his_results_trend.png")
FORECAST_PLOT = os.path.join(BASE_DIR, "his_future_forecast.png")
HEATMAP_PLOT = os.path.join(BASE_DIR, "his_revenue_heatmap.png")

CONFIG_PATH = os.path.join(BASE_DIR, "his_config.yaml")

# ===============================
# LOAD & CLEAN DATA
# ===============================
df = pd.read_csv(DATA_PATH)
df = df.loc[:, ~df.columns.str.contains("^Unnamed", na=False)]
df = df.iloc[:, :3]
df.columns = ["city", "year", "revenue"]

df["year"] = df["year"].astype(str).str.extract(r"(\d{4})")[0]
df["year"] = pd.to_numeric(df["year"], errors="coerce")
df = df.dropna(subset=["year", "revenue"])
df["year"] = df["year"].astype(int)

df = df.sort_values(["city", "year"])

# ===============================
# HEATMAP (SHOW + SAVE)
# ===============================
pivot = df.pivot_table(
    values="revenue",
    index="city",
    columns="year",
    aggfunc="sum"
)

plt.figure(figsize=(14, 8))
sns.heatmap(pivot, cmap="YlGnBu")
plt.title("HIS (AIS + HSA) | City-wise Tax Revenue Heatmap")
plt.tight_layout()
plt.show()
plt.savefig(HEATMAP_PLOT)
plt.close()

# ===============================
# YEARLY AGGREGATION
# ===============================
yearly = df.groupby("year", as_index=False)["revenue"].sum()

X = yearly["year"].values.reshape(-1, 1)
y = yearly["revenue"].values.reshape(-1, 1)

# ===============================
# SCALE TARGET
# ===============================
scaler = MinMaxScaler()
y_scaled = scaler.fit_transform(y)

# ===============================
# FITNESS FUNCTION
# ===============================
def fitness(params):
    w, b = params
    y_pred = w * X + b
    return mean_squared_error(y_scaled, y_pred)

# ===============================
# AIS INITIAL POPULATION
# ===============================
np.random.seed(42)
POP_SIZE = 20
ais_population = np.random.uniform(-1, 1, size=(POP_SIZE, 2))
ais_fitness = np.array([fitness(ind) for ind in ais_population])

# ===============================
# AIS CLONAL SELECTION
# ===============================
for _ in range(20):
    best_idx = np.argsort(ais_fitness)[:5]
    clones = ais_population[best_idx] + np.random.normal(0, 0.1, (5, 2))
    clone_fitness = np.array([fitness(c) for c in clones])

    combined = np.vstack([ais_population, clones])
    combined_fit = np.hstack([ais_fitness, clone_fitness])

    top = np.argsort(combined_fit)[:POP_SIZE]
    ais_population = combined[top]
    ais_fitness = combined_fit[top]

# ===============================
# HSA PARAMETERS
# ===============================
HM_SIZE = POP_SIZE
HMCR = 0.9     # Harmony Memory Considering Rate
PAR = 0.3      # Pitch Adjusting Rate
BW = 0.05      # Bandwidth
ITERATIONS = 30

harmony_memory = ais_population.copy()
harmony_fitness = np.array([fitness(h) for h in harmony_memory])

# ===============================
# HSA OPTIMIZATION LOOP
# ===============================
for _ in range(ITERATIONS):
    new_harmony = []

    for d in range(2):  # w and b
        if np.random.rand() < HMCR:
            value = harmony_memory[
                np.random.randint(0, HM_SIZE), d
            ]
            if np.random.rand() < PAR:
                value += np.random.uniform(-BW, BW)
        else:
            value = np.random.uniform(-1, 1)

        new_harmony.append(value)

    new_harmony = np.array(new_harmony)
    new_fitness = fitness(new_harmony)

    worst_idx = np.argmax(harmony_fitness)
    if new_fitness < harmony_fitness[worst_idx]:
        harmony_memory[worst_idx] = new_harmony
        harmony_fitness[worst_idx] = new_fitness

# ===============================
# BEST SOLUTION
# ===============================
best_solution = harmony_memory[np.argmin(harmony_fitness)]
w_opt, b_opt = best_solution

# ===============================
# FINAL PREDICTIONS
# ===============================
y_pred_scaled = w_opt * X + b_opt
y_pred = scaler.inverse_transform(y_pred_scaled)

# ===============================
# RESULTS CSV
# ===============================
results_df = pd.DataFrame({
    "year": yearly["year"],
    "actual_revenue": y.flatten(),
    "predicted_revenue": y_pred.flatten()
})
results_df.to_csv(RESULT_CSV, index=False)

# ===============================
# ACTUAL VS PREDICTED GRAPH
# ===============================
plt.figure(figsize=(10, 5))
plt.plot(results_df["year"], results_df["actual_revenue"], label="Actual", marker="o")
plt.plot(results_df["year"], results_df["predicted_revenue"], label="HIS Predicted", marker="o")
plt.title("HIS (AIS + HSA) | Actual vs Predicted Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.legend()
plt.show()
plt.savefig(COMPARE_PLOT)
plt.close()

# ===============================
# RESULTS TREND GRAPH
# ===============================
plt.figure(figsize=(10, 5))
plt.plot(results_df["year"], results_df["predicted_revenue"], marker="o")
plt.title("HIS (AIS + HSA) | Revenue Trend")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.show()
plt.savefig(RESULT_PLOT)
plt.close()

# ===============================
# FUTURE FORECAST
# ===============================
future_years = np.arange(
    yearly["year"].max() + 1,
    yearly["year"].max() + 6
).reshape(-1, 1)

future_scaled = w_opt * future_years + b_opt
future_values = scaler.inverse_transform(future_scaled)

future_df = pd.DataFrame({
    "year": future_years.flatten(),
    "forecast_revenue": future_values.flatten()
})
future_df.to_csv(FUTURE_CSV, index=False)

# ===============================
# FUTURE FORECAST GRAPH
# ===============================
plt.figure(figsize=(8, 4))
plt.plot(future_df["year"], future_df["forecast_revenue"], marker="o")
plt.title("HIS (AIS + HSA) | Future Tax Revenue Forecast")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.show()
plt.savefig(FORECAST_PLOT)
plt.close()

# ===============================
# SAVE CONFIG
# ===============================
config = {
    "model": "Hybrid AIS + HSA Optimized Regression (HIS)",
    "population_size": POP_SIZE,
    "ais_iterations": 20,
    "hsa_iterations": ITERATIONS,
    "harmony_memory_size": HM_SIZE,
    "optimized_parameters": {
        "slope": float(w_opt),
        "intercept": float(b_opt)
    }
}

with open(CONFIG_PATH, "w") as f:
    yaml.dump(config, f)

print("✅ HIS (AIS + HSA) PIPELINE COMPLETED SUCCESSFULLY")
print("📁 All outputs saved with 'his_' prefix")
```
