```python
import os
import yaml
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error

# ===============================
# PATHS
# ===============================
BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Tax Revenue Forecasting System"
DATA_PATH = os.path.join(BASE_DIR, "D48-VATandGST_24.csv")

RESULT_CSV = os.path.join(BASE_DIR, "hso_model_results.csv")
FUTURE_CSV = os.path.join(BASE_DIR, "hso_future_predictions.csv")

COMPARE_PLOT = os.path.join(BASE_DIR, "hso_actual_vs_predicted.png")
RESULT_PLOT = os.path.join(BASE_DIR, "hso_results_trend.png")
FORECAST_PLOT = os.path.join(BASE_DIR, "hso_future_forecast.png")
HEATMAP_PLOT = os.path.join(BASE_DIR, "hso_revenue_heatmap.png")

CONFIG_PATH = os.path.join(BASE_DIR, "hso_config.yaml")

# ===============================
# LOAD & CLEAN DATA
# ===============================
df = pd.read_csv(DATA_PATH)
df = df.loc[:, ~df.columns.str.contains("^Unnamed", na=False)]
df = df.iloc[:, :3]
df.columns = ["city", "year", "revenue"]

df["year"] = df["year"].astype(str).str.extract(r"(\d{4})")[0]
df["year"] = pd.to_numeric(df["year"], errors="coerce")
df = df.dropna(subset=["year", "revenue"])
df["year"] = df["year"].astype(int)

df = df.sort_values(["city", "year"])

# ===============================
# HEATMAP (SHOW + SAVE)
# ===============================
pivot = df.pivot_table(
    values="revenue",
    index="city",
    columns="year",
    aggfunc="sum"
)

plt.figure(figsize=(14, 8))
sns.heatmap(pivot, cmap="YlGnBu")
plt.title("HSO (PSO + HSA) | City-wise Tax Revenue Heatmap")
plt.tight_layout()
plt.show()
plt.savefig(HEATMAP_PLOT)
plt.close()

# ===============================
# YEARLY AGGREGATION
# ===============================
yearly = df.groupby("year", as_index=False)["revenue"].sum()

X = yearly["year"].values.reshape(-1, 1)
y = yearly["revenue"].values.reshape(-1, 1)

# ===============================
# SCALE TARGET
# ===============================
scaler = MinMaxScaler()
y_scaled = scaler.fit_transform(y)

# ===============================
# FITNESS FUNCTION
# ===============================
def fitness(params):
    w, b = params
    y_pred = w * X + b
    return mean_squared_error(y_scaled, y_pred)

# ===============================
# PSO INITIALIZATION
# ===============================
np.random.seed(42)
POP_SIZE = 20

particles = np.random.uniform(-1, 1, size=(POP_SIZE, 2))
velocities = np.random.uniform(-0.1, 0.1, size=(POP_SIZE, 2))

pbest = particles.copy()
pbest_fit = np.array([fitness(p) for p in particles])

gbest = pbest[np.argmin(pbest_fit)]

w_inertia = 0.7
c1 = 1.5
c2 = 1.5

# ===============================
# PSO OPTIMIZATION
# ===============================
for _ in range(30):
    for i in range(POP_SIZE):
        r1, r2 = np.random.rand(), np.random.rand()
        velocities[i] = (
            w_inertia * velocities[i]
            + c1 * r1 * (pbest[i] - particles[i])
            + c2 * r2 * (gbest - particles[i])
        )
        particles[i] += velocities[i]

        fit = fitness(particles[i])
        if fit < pbest_fit[i]:
            pbest[i] = particles[i]
            pbest_fit[i] = fit

    gbest = pbest[np.argmin(pbest_fit)]

# ===============================
# HSA PARAMETERS
# ===============================
HM_SIZE = POP_SIZE
HMCR = 0.9     # Harmony Memory Considering Rate
PAR = 0.3      # Pitch Adjusting Rate
BW = 0.05      # Bandwidth
ITERATIONS = 30

harmony_memory = particles.copy()
harmony_fitness = np.array([fitness(h) for h in harmony_memory])

# ===============================
# HSA REFINEMENT
# ===============================
for _ in range(ITERATIONS):
    new_harmony = []

    for d in range(2):  # w, b
        if np.random.rand() < HMCR:
            value = harmony_memory[np.random.randint(0, HM_SIZE), d]
            if np.random.rand() < PAR:
                value += np.random.uniform(-BW, BW)
        else:
            value = np.random.uniform(-1, 1)

        new_harmony.append(value)

    new_harmony = np.array(new_harmony)
    new_fit = fitness(new_harmony)

    worst_idx = np.argmax(harmony_fitness)
    if new_fit < harmony_fitness[worst_idx]:
        harmony_memory[worst_idx] = new_harmony
        harmony_fitness[worst_idx] = new_fit

# ===============================
# BEST SOLUTION
# ===============================
best_solution = harmony_memory[np.argmin(harmony_fitness)]
w_opt, b_opt = best_solution

# ===============================
# FINAL PREDICTIONS
# ===============================
y_pred_scaled = w_opt * X + b_opt
y_pred = scaler.inverse_transform(y_pred_scaled)

# ===============================
# RESULTS CSV
# ===============================
results_df = pd.DataFrame({
    "year": yearly["year"],
    "actual_revenue": y.flatten(),
    "predicted_revenue": y_pred.flatten()
})
results_df.to_csv(RESULT_CSV, index=False)

# ===============================
# ACTUAL VS PREDICTED GRAPH
# ===============================
plt.figure(figsize=(10, 5))
plt.plot(results_df["year"], results_df["actual_revenue"], label="Actual", marker="o")
plt.plot(results_df["year"], results_df["predicted_revenue"], label="HSO Predicted", marker="o")
plt.title("HSO (PSO + HSA) | Actual vs Predicted Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.legend()
plt.show()
plt.savefig(COMPARE_PLOT)
plt.close()

# ===============================
# RESULTS TREND GRAPH
# ===============================
plt.figure(figsize=(10, 5))
plt.plot(results_df["year"], results_df["predicted_revenue"], marker="o")
plt.title("HSO (PSO + HSA) | Revenue Trend")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.show()
plt.savefig(RESULT_PLOT)
plt.close()

# ===============================
# FUTURE FORECAST
# ===============================
future_years = np.arange(
    yearly["year"].max() + 1,
    yearly["year"].max() + 6
).reshape(-1, 1)

future_scaled = w_opt * future_years + b_opt
future_values = scaler.inverse_transform(future_scaled)

future_df = pd.DataFrame({
    "year": future_years.flatten(),
    "forecast_revenue": future_values.flatten()
})
future_df.to_csv(FUTURE_CSV, index=False)

# ===============================
# FUTURE FORECAST GRAPH
# ===============================
plt.figure(figsize=(8, 4))
plt.plot(future_df["year"], future_df["forecast_revenue"], marker="o")
plt.title("HSO (PSO + HSA) | Future Tax Revenue Forecast")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.show()
plt.savefig(FORECAST_PLOT)
plt.close()

# ===============================
# SAVE CONFIG
# ===============================
config = {
    "model": "Hybrid PSO + HSA Optimized Regression (HSO)",
    "population_size": POP_SIZE,
    "pso_iterations": 30,
    "hsa_iterations": ITERATIONS,
    "optimized_parameters": {
        "slope": float(w_opt),
        "intercept": float(b_opt)
    }
}

with open(CONFIG_PATH, "w") as f:
    yaml.dump(config, f)

print("✅ HSO (PSO + HSA) PIPELINE COMPLETED SUCCESSFULLY")
print("📁 All outputs saved with 'hso_' prefix")
```
