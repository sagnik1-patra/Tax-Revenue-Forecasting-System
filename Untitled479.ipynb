```python
import os
import yaml
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error

# ===============================
# PATHS
# ===============================
BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Tax Revenue Forecasting System"
DATA_PATH = os.path.join(BASE_DIR, "D48-VATandGST_24.csv")

RESULT_CSV = os.path.join(BASE_DIR, "gis_model_results.csv")
FUTURE_CSV = os.path.join(BASE_DIR, "gis_future_predictions.csv")

COMPARE_PLOT = os.path.join(BASE_DIR, "gis_actual_vs_predicted.png")
RESULT_PLOT = os.path.join(BASE_DIR, "gis_results_trend.png")
FORECAST_PLOT = os.path.join(BASE_DIR, "gis_future_forecast.png")
HEATMAP_PLOT = os.path.join(BASE_DIR, "gis_revenue_heatmap.png")

CONFIG_PATH = os.path.join(BASE_DIR, "gis_config.yaml")

# ===============================
# LOAD & CLEAN DATA
# ===============================
df = pd.read_csv(DATA_PATH)
df = df.loc[:, ~df.columns.str.contains("^Unnamed", na=False)]
df = df.iloc[:, :3]
df.columns = ["city", "year", "revenue"]

df["year"] = df["year"].astype(str).str.extract(r"(\d{4})")[0]
df["year"] = pd.to_numeric(df["year"], errors="coerce")
df = df.dropna(subset=["year", "revenue"])
df["year"] = df["year"].astype(int)

df = df.sort_values(["city", "year"])

# ===============================
# HEATMAP (SHOW + SAVE)
# ===============================
pivot = df.pivot_table(
    values="revenue",
    index="city",
    columns="year",
    aggfunc="sum"
)

plt.figure(figsize=(14, 8))
sns.heatmap(pivot, cmap="YlGnBu")
plt.title("GIS (GA + AIS) | City-wise Tax Revenue Heatmap")
plt.tight_layout()
plt.show()
plt.savefig(HEATMAP_PLOT)
plt.close()

# ===============================
# YEARLY AGGREGATION
# ===============================
yearly = df.groupby("year", as_index=False)["revenue"].sum()

X = yearly["year"].values.reshape(-1, 1)
y = yearly["revenue"].values.reshape(-1, 1)

# ===============================
# SCALE TARGET
# ===============================
scaler = MinMaxScaler()
y_scaled = scaler.fit_transform(y)

# ===============================
# FITNESS FUNCTION
# ===============================
def fitness(individual):
    w, b = individual
    y_pred = w * X + b
    return mean_squared_error(y_scaled, y_pred)

# ===============================
# AIS INITIAL POPULATION
# ===============================
np.random.seed(42)
POP_SIZE = 20
population = np.random.uniform(-1, 1, size=(POP_SIZE, 2))

# ===============================
# GA + AIS LOOP
# ===============================
GENERATIONS = 40
MUTATION_RATE = 0.2

for _ in range(GENERATIONS):
    fitness_vals = np.array([fitness(ind) for ind in population])

    # AIS-style selection (top antibodies)
    elite_idx = np.argsort(fitness_vals)[:5]
    elites = population[elite_idx]

    # GA crossover
    offspring = []
    while len(offspring) < POP_SIZE - len(elites):
        p1, p2 = elites[np.random.randint(0, len(elites), 2)]
        alpha = np.random.rand()
        child = alpha * p1 + (1 - alpha) * p2

        # Mutation
        if np.random.rand() < MUTATION_RATE:
            child += np.random.normal(0, 0.1, 2)

        offspring.append(child)

    population = np.vstack([elites, offspring])

# ===============================
# BEST SOLUTION
# ===============================
fitness_vals = np.array([fitness(ind) for ind in population])
best_solution = population[np.argmin(fitness_vals)]
w_opt, b_opt = best_solution

# ===============================
# FINAL PREDICTIONS
# ===============================
y_pred_scaled = w_opt * X + b_opt
y_pred = scaler.inverse_transform(y_pred_scaled)

# ===============================
# RESULTS CSV
# ===============================
results_df = pd.DataFrame({
    "year": yearly["year"],
    "actual_revenue": y.flatten(),
    "predicted_revenue": y_pred.flatten()
})
results_df.to_csv(RESULT_CSV, index=False)

# ===============================
# ACTUAL VS PREDICTED GRAPH
# ===============================
plt.figure(figsize=(10, 5))
plt.plot(results_df["year"], results_df["actual_revenue"], label="Actual", marker="o")
plt.plot(results_df["year"], results_df["predicted_revenue"], label="GIS Predicted", marker="o")
plt.title("GIS (GA + AIS) | Actual vs Predicted Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.legend()
plt.show()
plt.savefig(COMPARE_PLOT)
plt.close()

# ===============================
# RESULTS TREND GRAPH
# ===============================
plt.figure(figsize=(10, 5))
plt.plot(results_df["year"], results_df["predicted_revenue"], marker="o")
plt.title("GIS (GA + AIS) | Revenue Trend")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.show()
plt.savefig(RESULT_PLOT)
plt.close()

# ===============================
# FUTURE FORECAST
# ===============================
future_years = np.arange(
    yearly["year"].max() + 1,
    yearly["year"].max() + 6
).reshape(-1, 1)

future_scaled = w_opt * future_years + b_opt
future_values = scaler.inverse_transform(future_scaled)

future_df = pd.DataFrame({
    "year": future_years.flatten(),
    "forecast_revenue": future_values.flatten()
})
future_df.to_csv(FUTURE_CSV, index=False)

# ===============================
# FUTURE FORECAST GRAPH
# ===============================
plt.figure(figsize=(8, 4))
plt.plot(future_df["year"], future_df["forecast_revenue"], marker="o")
plt.title("GIS (GA + AIS) | Future Tax Revenue Forecast")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.show()
plt.savefig(FORECAST_PLOT)
plt.close()

# ===============================
# SAVE CONFIG
# ===============================
config = {
    "model": "Hybrid GA + AIS Optimized Regression (GIS)",
    "population_size": POP_SIZE,
    "generations": GENERATIONS,
    "mutation_rate": MUTATION_RATE,
    "optimized_parameters": {
        "slope": float(w_opt),
        "intercept": float(b_opt)
    }
}

with open(CONFIG_PATH, "w") as f:
    yaml.dump(config, f)

print("✅ GIS (GA + AIS) PIPELINE COMPLETED SUCCESSFULLY")
print("📁 All outputs saved with 'gis_' prefix")
```
