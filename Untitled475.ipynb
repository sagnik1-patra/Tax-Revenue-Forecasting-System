```python
import os
import yaml
import joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import MinMaxScaler

# ===============================
# PATHS
# ===============================
BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Tax Revenue Forecasting System"
DATA_PATH = os.path.join(BASE_DIR, "D48-VATandGST_24.csv")

SCALER_PATH = os.path.join(BASE_DIR, "scaler.pkl")
CONFIG_PATH = os.path.join(BASE_DIR, "config.yaml")

RESULT_CSV = os.path.join(BASE_DIR, "model_results.csv")
FUTURE_CSV = os.path.join(BASE_DIR, "future_predictions.csv")

COMPARE_PLOT = os.path.join(BASE_DIR, "actual_vs_predicted.png")
HEATMAP_PLOT = os.path.join(BASE_DIR, "revenue_heatmap.png")
RESULT_PLOT = os.path.join(BASE_DIR, "results_trend.png")
FORECAST_PLOT = os.path.join(BASE_DIR, "future_forecast.png")

# ===============================
# LOAD & CLEAN DATA
# ===============================
df = pd.read_csv(DATA_PATH)
df = df.loc[:, ~df.columns.str.contains("^Unnamed", na=False)]
df = df.iloc[:, :3]
df.columns = ["city", "year", "revenue"]

df["year"] = (
    df["year"].astype(str)
    .str.extract(r"(\d{4})")[0]
)
df["year"] = pd.to_numeric(df["year"], errors="coerce")
df = df.dropna(subset=["year", "revenue"])
df["year"] = df["year"].astype(int)

df = df.sort_values(["city", "year"])

# ===============================
# HEATMAP (SHOW ➜ SAVE)
# ===============================
pivot = df.pivot_table(values="revenue", index="city", columns="year", aggfunc="sum")

plt.figure(figsize=(14, 8))
sns.heatmap(pivot, cmap="YlGnBu")
plt.title("City-wise Tax Revenue Heatmap")
plt.tight_layout()
plt.show()                      # 👈 SHOW
plt.savefig(HEATMAP_PLOT)       # 👈 SAVE
plt.close()

# ===============================
# YEARLY AGGREGATION
# ===============================
yearly = df.groupby("year", as_index=False)["revenue"].sum()

X = yearly[["year"]]
y = yearly["revenue"]

# ===============================
# SCALING
# ===============================
scaler = MinMaxScaler()
y_scaled = scaler.fit_transform(y.values.reshape(-1, 1))
joblib.dump(scaler, SCALER_PATH)

# ===============================
# MODEL
# ===============================
model = LinearRegression()
model.fit(X, y_scaled)

# ===============================
# PREDICTIONS
# ===============================
y_pred_scaled = model.predict(X)
y_pred = scaler.inverse_transform(y_pred_scaled)

# ===============================
# RESULTS CSV
# ===============================
results_df = pd.DataFrame({
    "year": yearly["year"],
    "actual_revenue": y.values,
    "predicted_revenue": y_pred.flatten()
})
results_df.to_csv(RESULT_CSV, index=False)

# ===============================
# ACTUAL VS PREDICTED GRAPH
# ===============================
plt.figure(figsize=(10, 5))
plt.plot(results_df["year"], results_df["actual_revenue"], label="Actual", marker="o")
plt.plot(results_df["year"], results_df["predicted_revenue"], label="Predicted", marker="o")
plt.title("Actual vs Predicted Tax Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.legend()
plt.show()                       # 👈 SHOW
plt.savefig(COMPARE_PLOT)        # 👈 SAVE
plt.close()

# ===============================
# RESULTS TREND GRAPH
# ===============================
plt.figure(figsize=(10, 5))
plt.plot(results_df["year"], results_df["predicted_revenue"], marker="o")
plt.title("Predicted Revenue Trend")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.show()                       # 👈 SHOW
plt.savefig(RESULT_PLOT)         # 👈 SAVE
plt.close()

# ===============================
# FUTURE FORECAST
# ===============================
future_years = np.arange(
    yearly["year"].max() + 1,
    yearly["year"].max() + 6
).reshape(-1, 1)

future_scaled = model.predict(future_years)
future_values = scaler.inverse_transform(future_scaled)

future_df = pd.DataFrame({
    "year": future_years.flatten(),
    "forecast_revenue": future_values.flatten()
})
future_df.to_csv(FUTURE_CSV, index=False)

# ===============================
# FUTURE FORECAST GRAPH
# ===============================
plt.figure(figsize=(8, 4))
plt.plot(future_df["year"], future_df["forecast_revenue"], marker="o")
plt.title("Future Tax Revenue Forecast")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.show()                       # 👈 SHOW
plt.savefig(FORECAST_PLOT)       # 👈 SAVE
plt.close()

# ===============================
# SAVE CONFIG
# ===============================
config = {
    "model": "Linear Regression",
    "dataset": "D48-VATandGST_24.csv",
    "forecast_years": 5,
    "note": "Graphs displayed before saving"
}

with open(CONFIG_PATH, "w") as f:
    yaml.dump(config, f)

print("✅ ALL GRAPHS SHOWN ON SCREEN AND SAVED SUCCESSFULLY")
print("📁 Folder:", BASE_DIR)
```
